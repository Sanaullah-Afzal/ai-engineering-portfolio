### Gemini API Check

```python
import os #manage env virables
from langchain.chat_models import init_chat_model # library used to initialize any kind of chat models
# using init chat model gives you an option of indirectly using ChatOpenai
# os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")
# No need to reassign GEMINI_API_KEY if it's already in environment
model = init_chat_model("google_genai:gemini-3.5-flash") #it work with google_genai:

response = model.invoke("How are you?")
print(response.content)
```

### Groq API without init_chat_model, direct approach

```python
# Another way of doing WIH groqchat INSTEAD OF init_chat_model
import os

from langchain_groq import ChatGroq
# load_dotenv()  # not needed if already set in environment, picks up GEMINI_API_KEY from environment
# os.environ["GEMINI_API_KEY"]=os.getenv("GEMINI_API_KEY")

model = ChatGroq(model="openai/gpt-oss-120b") #works with openai/, not with openai:

response = model.invoke("How are you?")
print(response.content)
```

## Streaming and Batch

### Streaming

```python
model.invoke("Generate 200 words para on Abbottabad")
```

```python
for chunk in model.stream("Generate 200 words para on Abbottabad"):
    print(chunk.content, end="|", flush=True) # | delimeter to show the end of each chunk, flush=True to force immediate output
    # print(chunk.text) # does not displays output
```

### Batch
 Batching refers to running multiple inputs through your graph (or agent workflow) at once, instead of invoking it one by one. This is useful when you want to process a list of prompts, documents, or tasks in parallel or sequentially with shared sta

```python
responses = model.batch([
    "Generate 200 words para on Abbottabad",
    "Generate 200 words para on Islamabad",
    "Generate 200 words para on Lahore"
])

for response in responses:
    print(response.content)
```

```python
responses = model.batch([
    "Generate 200 words para on Abbottabad",
    "Generate 200 words para on Islamabad",
    "Generate 200 words para on Lahore"
],
config={'max_concurrent_requests': 3} #limit parallel request to 3
)

for response in responses:
    print(response.content)
```
